## Generate Images with Stable Diffusion

Stable Diffusion is a popular open-source model that can generate high-quality images from text prompts. We'll use the Hugging Face `diffusers` library for this.

```python
# Install necessary libraries for Stable Diffusion
!pip install -q diffusers transformers accelerate scipy

print("Stable Diffusion libraries installed.")
```

Now, let's load a Stable Diffusion pipeline. We'll use the `StableDiffusionPipeline` and a base model like `runwayml/stable-diffusion-v1-5` for demonstration. For better quality, consider `stabilityai/stable-diffusion-xl-base-1.0` if you have sufficient GPU resources and accept the larger model size.

```python
import torch
from diffusers import StableDiffusionPipeline
from IPython.display import Image, display

# Load the Stable Diffusion pipeline
# You can choose a different model if you prefer, e.g., 'stabilityai/stable-diffusion-xl-base-1.0'
# Note: Larger models require more VRAM and can take longer to load and generate.
model_id = "runwayml/stable-diffusion-v1-5"
pipeline = StableDiffusionPipeline.from_pretrained(model_id, torch_dtype=torch.float16)
pipeline.to("cuda")

print(f"Stable Diffusion pipeline '{model_id}' loaded successfully.")
```

With the pipeline loaded, you can now generate images by providing a text prompt. You can adjust the prompt and other parameters to get different results.

```python
#@title Generate an Image with Stable Diffusion

# Define your image generation prompt
sd_prompt = "Dog, with sunset scenery on the background, cinematic view " # @param {type:"string"}

# Generate the image
with torch.no_grad():
    image = pipeline(sd_prompt).images[0]

# Display the generated image
display(image)

print(f"Image generated for prompt: '{sd_prompt}'")
```
